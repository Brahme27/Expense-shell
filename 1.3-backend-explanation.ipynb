Here are the **backend script main steps** for quick notes 👇


## 📝 Backend Setup Steps

1. Check root user
2. Setup logging
3. Install NodeJS (disable old → enable v20 → install)
4. Create `expense` user (if not exists)
5. Create `/app` directory
6. Download backend code (zip file)
7. Clean `/app` folder
8. Unzip backend code
9. Install NodeJS dependencies (`npm install`)
10. Copy backend service file
11. Install MySQL client
12. Load database schema (tables)
13. Reload system services
14. Enable backend service
15. Start backend service

**block by block in simple words** 👇

# 🧱 1. Basic setup

```bash
#!/bin/bash

USERID=$(id -u)
```

👉 Runs script in bash
👉 Gets user ID (0 = root)

---

# 🎨 2. Colors

```bash
R="\e[31m"
G="\e[32m"
Y="\e[33m"
N="\e[0m"
```

👉 Used to print:

* Red → failure
* Green → success
* Yellow → warning

---

# 📁 3. Log setup

```bash
LOGS_FOLDER="/var/log/expense-logs"
LOG_FILE=$(echo $0 | cut -d "." -f1 )
TIMESTAMP=$(date +%Y-%m-%d-%H-%M-%S)
LOG_FILE_NAME="$LOGS_FOLDER/$LOG_FILE-$TIMESTAMP.log"
```

👉 Creates log file like:

```
/var/log/expense-logs/scriptname-2026-04-25.log
```

---

# ⚙️ 4. VALIDATE function

```bash
VALIDATE(){
    if [ $1 -ne 0 ]
```

👉 Checks if command failed

* `$1` → exit code
* `$2` → message

👉 If fail → stop script
👉 If success → print success

---

# 🔐 5. CHECK_ROOT function

```bash
CHECK_ROOT(){
    if [ $USERID -ne 0 ]
```

👉 Ensures script runs as root
👉 Stops if not root

---

# 🧾 6. Start logging

```bash
echo "Script started executing at: $TIMESTAMP" &>>$LOG_FILE_NAME
```

👉 Writes start message into log file

---

# 🔐 7. Root check

```bash
CHECK_ROOT
```

👉 Stops script if not sudo/root

---

# 📦 8. Install NodeJS

```bash
dnf module disable nodejs -y
VALIDATE $? "Disabling existing default NodeJS"
```

👉 Removes old NodeJS version

---

```bash
dnf module enable nodejs:20 -y
VALIDATE $? "Enabling NodeJS 20"
```

👉 Enables NodeJS version 20

---

```bash
dnf install nodejs -y
VALIDATE $? "Installing NodeJS"
```

👉 Installs NodeJS

---

# 👤 9. Create user

```bash
id expense
```

👉 Checks if user exists

---

```bash
if [ $? -ne 0 ]
then
    useradd expense
```

👉 If not exists → create user

---

```bash
else
    echo "already exists"
```

👉 If exists → skip

---

# 📁 10. Create app directory

```bash
mkdir -p /app
VALIDATE $? "Creating app directory"
```

👉 Creates `/app` safely
👉 No error if already exists

---

# 🌐 11. Download backend code

```bash
curl -o /tmp/backend.zip https://expense-builds.s3...
```

👉 Downloads backend zip file

---

# 📂 12. Prepare app folder

```bash
cd /app
rm -rf /app/*
```

👉 Go to `/app`
👉 Remove old files

---

# 📦 13. Extract backend

```bash
unzip /tmp/backend.zip
VALIDATE $? "unzip backend"
```

👉 Extracts backend code

---

# 📦 14. Install dependencies

```bash
npm install
VALIDATE $? "Installing dependencies"
```

👉 Installs NodeJS packages

---

# ⚙️ 15. Copy service file

```bash
cp /home/ec2-user/expense-shell/backend.service /etc/systemd/system/backend.service
```

👉 Copies service file
👉 Used to run backend as a service

---

# 🗄️ 16. Install MySQL client

```bash
dnf install mysql -y
VALIDATE $? "Installing MySQL Client"
```

👉 Needed to connect to database

---

# 🧾 17. Setup database

```bash
mysql -h mysql.brahme.me -uroot -pExpenseApp@1 < /app/schema/backend.sql
```

👉 Runs SQL file
👉 Creates tables & schema

---

# 🔄 18. Reload systemd

```bash
systemctl daemon-reload
VALIDATE $? "Daemon Reload"
```

👉 Reloads service configs

---

# 🚀 19. Enable backend

```bash
systemctl enable backend
VALIDATE $? "Enabling backend"
```

👉 Starts backend on boot

---

# ▶️ 20. Start backend

```bash
systemctl restart backend
VALIDATE $? "Starting Backend"
```

👉 Starts backend service

---
